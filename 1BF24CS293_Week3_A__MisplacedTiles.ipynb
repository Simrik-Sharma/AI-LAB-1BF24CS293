```python
import heapq
initial_state = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)
goal_state = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)
def display(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()
def misplaced_tiles(state):
    count = 0
    for i in range(9):
        if state[i] != 0 and state[i] != goal_state[i]:
            count += 1
    return count
def get_neighbors(state):
    neighbors = []
    blank_index = state.index(0)
    row = blank_index // 3
    col = blank_index % 3
    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]
    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_index = new_row * 3 + new_col
            new_state = list(state)
            new_state[blank_index], new_state[new_index] = \
                new_state[new_index], new_state[blank_index]
            neighbors.append(tuple(new_state))
    return neighbors
def a_star_misplaced():
    priority_queue = []
    h = misplaced_tiles(initial_state)
    heapq.heappush(
        priority_queue,
        (h, 0, initial_state)
    )
    g_cost = {
        initial_state: 0
    }
    parent = {
        initial_state: None
    }
    while priority_queue:
        f, g, current = heapq.heappop(priority_queue)
        if current == goal_state:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]
            path.reverse()
            return path
        for neighbor in get_neighbors(current):
            new_g = g + 1
            if neighbor not in g_cost or new_g < g_cost[neighbor]:
                g_cost[neighbor] = new_g
                parent[neighbor] = current
                h = misplaced_tiles(neighbor)
                f = new_g + h
                heapq.heappush(
                    priority_queue,
                    (f, new_g, neighbor)
                )
    return None
print("A* SEARCH USING MISPLACED TILES\n")
print("Initial State:")
display(initial_state)
print("Goal State:")
display(goal_state)
print("Initial h(n) =", misplaced_tiles(initial_state))
path = a_star_misplaced()
if path:
    print("Solution found!")
    print("Number of moves =", len(path) - 1)
    print("\nSolution Path:\n")
    for i, state in enumerate(path):
        print("Step", i)
        display(state)
else:
    print("No solution found.")
```
